# Grafico de Dispersion (Scatter plots)

El grafico de dispersion, es una representacion grafica que muestra como se relacionan dos variables cuantitativas.

Cada uno de los puntos del grafico representa un par de valores $(X,Y)$ para un solo caso u observacion. Donde:
- $X$ es la variable independiente o explicativa.
- $Y$ es la variable dependiente.

Ademas, los graficos de dispersion sirven, para visualizar relaciones, identificar correlaciones y detectar patrones como tendencias, agrupaciones o valores atípicos (outliers).


Para poder hacer nuestros primeros graficos de dispersion, generaremos datos para tener una mejor visualizacion y una explicacion mas clara de este grafico:

Una de las cosas que hacemos al tener un grafico de dispersion es lo siguiente:

**Revisar la direccion o tendencia:**

- **Positiva:** Conforme X aumenta, Y tiende a aumentar.
- **Negativa:** Conforme X aumenta, Y tienede a disminuir.
- **Nula:** No se aprecia una relacion clara entre el cambio de X y el cambio de Y

**Forma:**
- **Lineal:** La nueve de puntos sigue aproximadamente una trayectoria rectilinea.
- **No lineal:** La relacionn sigue patrones cuadraticos
- **Valores atipicos (outliers):** Observaciones que se alejan significativemente del patron general que muestra el resto de la nube de puntos.

## Linea de dispersion

La línea de dispersión, formalmente conocida como línea de regresión, es una línea que se dibuja sobre un diagrama de dispersión para representar visualmente el comportamiento general, la dirección y la fuerza de la relación entre dos variables.

**¿Cómo funciona?**

La recta tiene la forma:

**ŷ = β₀ + β₁x**

β₀ (intercepto): el valor estimado de Y cuando X = 0.

β₁ (pendiente): cuánto cambia Y en promedio por cada unidad que aumenta X.

ŷ: el valor de Y que predice la recta para un X dado.

Para elegir cuál recta es la mejor se usa el método de mínimos cuadrados ordinarios (MCO). Para cada punto se mide la distancia vertical a la recta (el residuo, eᵢ = yᵢ − ŷᵢ), se elevan al cuadrado y se busca la recta que minimiza la suma de esos cuadrados:

min Σ(yᵢ − ŷᵢ)²

Elevar al cuadrado evita que los residuos positivos y negativos se cancelen y penaliza más los errores grandes.

**¿Para qué sirve?**

Describir la relación: el signo de la pendiente indica si es positiva o negativa, y su magnitud qué tan fuerte es el efecto.

Predecir: dado un valor de X, estimar el valor esperado de Y.

Cuantificar el ajuste: con el R² (proporción de la variabilidad de Y explicada por X) y el coeficiente de correlación.

Inferencia: probar si la relación es estadísticamente significativa (prueba t sobre β₁, intervalos de confianza).

Detectar problemas: al graficar los residuos se ven valores atípicos, no linealidad o varianza no constante (heterocedasticidad).


**Ojo**

Que la línea se ajuste perfectamente a los puntos indica una alta correlación, pero no significa automáticamente que una variable sea la causa de la otra.

# **4. Mapa de Calor (Heatmap) y Matriz de Correlación**

El **mapa de calor** es una representación gráfica de datos tridimensionales o matrices donde los valores individuales contenidos en una matriz se representan mediante colores.

En el análisis de datos, se utiliza principalmente para visualizar la **Matriz de Correlación de Pearson** entre variables cuantitativas continuas.

## Coeficiente de Correlación de Pearson ($r$)
El coeficiente de correlación de Pearson mide la fuerza y dirección de la relación lineal entre dos variables aleatorias $X$ e $Y$:

$$r = \frac{\text{Cov}(X,Y)}{\sigma_X \sigma_Y} = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n} (x_i - \bar{x})^2} \sqrt{\sum_{i=1}^{n} (y_i - \bar{y})^2}}$$

### Propiedades del Coeficiente $r$:
* **$r = 1$:** Correlación positiva perfecta.
* **$r = -1$:** Correlación negativa perfecta.
* **$r = 0$:** Ausencia de relación lineal.
* **Rango:** $-1 \le r \le 1$.

**Para calcular el coeficiente de correlación de pearson hay diferentes formas**

```python
# con Pandas

r = df_reducido["age"].corr(df_reducido["bmi"], method="pearson")
print(f"r = {r:.4f}")

r2= df_reducido["age"].corr(df_reducido["body_fat_percentage"], method="pearson")
print(f"r2 = {r2:.4f}")
```

si los datos tienen valores faltantes, pandas los ignora automáticamente, pero pearsonr y np.corrcoef no

**primero se tiene que utilizar df.dropna**

```python
# Con scipy (incluye valor p)
from scipy.stats import pearsonr

datos = df_reducido[["age", "bmi", "body_fat_percentage"]].dropna() #quitamos valores faltantes

r3, p3 = pearsonr(datos["age"], datos["bmi"])
print(f"r = {r3:.4f}")
print(f"valor p = {p3:.4f}")

r4, p4 = pearsonr(datos["age"], datos["body_fat_percentage"])
print(f"r = {r4:.4f}")
print(f"valor p = {p4:.4f}")
```

```python
#con numpy

r5 = np.corrcoef(datos["age"], datos["bmi"])[0, 1] #devuelve una matriz 2×2, por eso se pone [0, 1].
print(f"r = {r5:.4f}")

r6 = np.corrcoef(datos["age"], datos["body_fat_percentage"])[0, 1]
print(f"r = {r6:.4f}")
```

## Mapa De Calor Basico

Para realizar el mapa de calor básico, usaremos las librerías `Seaborn` y `Matplotlib`.

Para el mapa de calor es de vital importancia que usemos la función `sns.heatmap()`, donde los parámetros más importantes de esta función son:

- `data`: El DataFrame, matriz o tabla de datos numéricos que se representará visualmente. **Es el único parámetro obligatorio.**
- `annot`: Un valor booleano (`True` o `False`). Si se define como `True`, muestra el valor numérico en el centro de cada celda.
- `fmt`: Define el formato numérico del texto cuando `annot=True` (por ejemplo, `".2f"` para 2 decimales o `"d"` para enteros).
- `cmap`: Especifica la paleta de colores para mapear los valores numéricos (por ejemplo, `"YlOrRd"`, `"coolwarm"`, `"viridis"`, `"magma"`).
- `vmin` y `vmax`: Establecen los límites inferior y superior de la escala de color. En matrices de correlación se suele usar `vmin=-1` y `vmax=1`.
- `linewidths`: Controla el ancho de las líneas divisorias que separan visualmente cada celda (por ejemplo, `linewidths=0.5`).

**IMPORTANTE:**
- El parametro data es un DataFrame que contiene las correlaciones de cada una de las variables, para esto usaremos la funcion `df[cols].corr()`
- - Este gráfico requiere trabajar exclusivamente con variables numéricas. En caso de contar con variables categóricas de tipo texto (*strings*), es necesario codificarlas o convertirlas previamente a valores numéricos (por ejemplo, mediante codificación ordinal o variables *dummy*)

```python

# 1. Seleccionar solo las variables numéricas continuas
cols_numericas = [
    "age",
    "bmi",
    "waist_to_hip_ratio",
    "body_fat_percentage",
    "fasting_glucose_mg_dl",
    "hba1c_percent",
    "fasting_insulin_uIU_ml",
    "insulin_resistance_index",
    "systolic_bp",
    "diastolic_bp",
    "total_cholesterol_mg_dl",
    "ldl_mg_dl",
    "hdl_mg_dl",
    "triglycerides_mg_dl",
]

# 2. Calcular la matriz de correlación de Pearson
matriz_corr = df_reducido[cols_numericas].corr()

# 3. Graficar el mapa de calor
plt.figure(figsize=(12, 10))
sns.heatmap(
    matriz_corr,
    annot=True,
    fmt=".2f",
    cmap="inferno",
    vmin=-1,
    vmax=1,
    linewidths=0.5,
)
plt.title("Matriz de Correlación de Variables Clínicas y Antropométricas")
plt.tight_layout() #Ajusta los margenes del grafico
plt.show()
```

Este grafico nos ayuda a ver de forma visual las correlaciones que existe entre variables, que tan correlacionado esta una variable de otra.
